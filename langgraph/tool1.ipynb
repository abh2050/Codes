```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
```

```python
api_wrapper = WikipediaAPIWrapper(top_k_results=5,doc_content_char_max_length=1000)
wikipedia_tool = WikipediaQueryRun(api_wrapper=api_wrapper)
```

```python
wikipedia_tool.name
```

```python
wikipedia_tool.description
```

```python
wikipedia_tool.args
```

```python
wikipedia_tool.run({"query": "who is Elon musk the engineer?"})
```

```python
wikipedia_tool.run('rcb')
```

## Building a custom Tool

```python
#to convert it to langchain tool
from langchain.agents import tool
```

```python
@tool
def mutiply(a:int,b:int) -> int:
    """Multiplies two numbers.""" #doc string is mandatory
    return a * b
```

```python
mutiply.invoke({"a": 2, "b": 3})
```

```python
mutiply.args
```

```python
mutiply.description
```

```python
@tool
def get_word_length(word: str) -> int:
    """Returns the length of the word."""
    return len(word)
```

```python
get_word_length.invoke({"word": "hello"})
```
